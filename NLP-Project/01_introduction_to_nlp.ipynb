## What is NLP?

**Natural Language Processing (NLP)** is the branch of AI that allows computers to read, understand, and generate human language.

| Application | Where you've seen it |
|---|---|
| Search engines | Google, Bing |
| Voice assistants | Siri, Alexa, Google Assistant |
| Translation | Google Translate, DeepL |
| Spam filters | Gmail, Outlook |
| Chatbots | ChatGPT, customer-support bots |
| Auto-correct | Your phone keyboard |

## NLTK vs spaCy

| | **NLTK** | **spaCy** |
|---|---|---|
| Style | Educational, modular | Production-ready |
| Speed | Slower | Very fast (Cython) |
| Models | Mostly rule-based | Pre-trained statistical models |
| Best for | Learning concepts | Real applications |

## Step 2 — Our Sample Text
Let's pretend this is a tiny news article that we want the computer to *understand*.

````python
text = (
    'Natural Language Processing is an exciting field of Artificial Intelligence. '
    'Companies like Google, Microsoft, and OpenAI are investing heavily in NLP. '
    'In 2023, ChatGPT became one of the fastest-growing apps in the world.'
)
print(text)
````

## Part A — NLTK

### A.1  Sentence Tokenization
Split a paragraph into individual sentences.

````python
from nltk.tokenize import sent_tokenize, word_tokenize

sentences = sent_tokenize(text)
for i, s in enumerate(sentences, 1):
    print(f'{i}. {s}')
````

### A.2  Word Tokenization
Split a sentence into individual *tokens* (words + punctuation).

````python
words = word_tokenize(text)
print('First 15 tokens:', words[:15])
````

### A.3  Part-of-Speech (POS) Tagging
Label each word with its grammatical role.

Common tags: `NN` = noun, `VB` = verb, `JJ` = adjective, `IN` = preposition, `DT` = determiner.

````python
pos_tags = nltk.pos_tag(words)
print(pos_tags[:10])
````

### A.4  Named Entity Recognition (NER)
Find real-world entities — people, organizations, dates, locations, money…

````python
entities = nltk.ne_chunk(pos_tags)
for chunk in entities:
    if hasattr(chunk, 'label'):
        name = ' '.join(c[0] for c in chunk)
        print(f'  {name}  ({chunk.label()})')
````

## Part B — spaCy

spaCy gives you tokenization, POS, lemmatization, NER, and dependency parsing **all in one call**.

````python
doc = nlp(text)
type(doc), len(doc)
````

### B.1  Sentences

````python
for i, sent in enumerate(doc.sents, 1):
    print(f'{i}. {sent.text}')
````

### B.2  Tokens + POS
Each spaCy token is a rich object with many attributes:
- `token.text`  — the original word
- `token.pos_`  — coarse POS (NOUN, VERB, ADJ…)
- `token.tag_`  — fine-grained tag (NN, NNS, VBD…)
- `token.lemma_` — dictionary form

````python
for token in doc[:10]:
    print(f'{token.text:<15} POS={token.pos_:<8} TAG={token.tag_}')
````

### B.3  Named Entities (much cleaner than NLTK)

Entity labels include: `PERSON`, `ORG`, `GPE` (countries/cities), `DATE`, `MONEY`, `PRODUCT`.

````python
for ent in doc.ents:
    print(f'  {ent.text:<20} ({ent.label_})')
````

### B.4  Lemmatization
Reduce each word to its dictionary base form: *running → run, better → good*.

````python
for token in doc[:10]:
    print(f'{token.text:<15} → {token.lemma_}')
````

## Summary

You just performed the four most common NLP tasks:
1. **Tokenization** — splitting text into words/sentences
2. **POS Tagging** — finding grammatical roles
3. **NER** — identifying real-world entities
4. **Lemmatization** — reducing words to their root form

**Rule of thumb:** use NLTK to learn, spaCy to ship.

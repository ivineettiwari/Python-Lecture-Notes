## 📊 Dataset Description

The dataset contains information about residential properties in Mumbai with the following columns:

- **bhk**: Number of bedrooms/hall/kitchen

- **locality**: Specific location/area in Mumbai
- **area**: Property area in square feet
- **price**: Property price
- **price_unit**: Currency unit (Lakhs 'L' or Crores 'Cr')
- **region**: Broader geographical region in Mumbai
- **status**: Construction status (Ready to move, Under Construction)
- **age**: Property age (New, Resale)

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# Set style for better visualizations
plt.style.use('seaborn-v0_8')
sns.set_palette(["#2E86AB", "#A23B72", "#F18F01", "#C73E1D"]) 

print("📊 MUMBAI HOUSE PRICES ANALYSIS PROJECT")
print("=" * 50)
````

````python
# Load the dataset
df = pd.read_csv('data/Mumbai House Prices.csv')

print("🔍 INITIAL DATA EXPLORATION")
print("=" * 40)

# Basic information about the dataset
print(f"Dataset Shape: {df.shape}")
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
print("\nColumn Names:")
print(df.columns.tolist())
print("\n" + "="*40)

# Display first few rows
print("First 5 rows of the dataset:")
print(df.head())
print("\n" + "="*40)

# Data types and basic info
print("Data Types and Info:")
print(df.info())
print("\n" + "="*40)

# Basic statistics
print("Basic Statistical Summary:")
print(df.describe())
print("\n" + "="*40)
````

````python
print("🧹 DATA CLEANING AND PREPROCESSING")
print("=" * 40)

# Check for missing values
print("Missing Values Analysis:")
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
print("\n" + "="*40)

# Check for duplicates
print(f"Number of duplicate rows: {df.duplicated().sum()}")

# Remove duplicates if any
df_clean = df.drop_duplicates()
print(f"Shape after removing duplicates: {df_clean.shape}")
print("\n" + "="*40)

# Check unique values in categorical columns
categorical_cols = ['type', 'locality', 'price_unit', 'region', 'status', 'age']
print("Unique values in categorical columns:")
for col in categorical_cols:
    print(f"{col}: {df_clean[col].nunique()} unique values")
print("\n" + "="*40)

# Handle price_unit column - convert all prices to a common unit (Lakhs)
print("Standardizing Price Units...")

def standardize_price(row):
    """
    Convert all prices to Lakhs for consistency
    Cr (Crore) = 100 Lakhs
    """
    if row['price_unit'] == 'Cr':
        return row['price'] * 100  # Convert Crore to Lakhs
    else:  # 'L' for Lakhs
        return row['price']

# Apply standardization
df_clean['price_lakhs'] = df_clean.apply(standardize_price, axis=1)
print("Price standardization completed!")
print("\n" + "="*40)

# Create price per sqft feature
df_clean['price_per_sqft'] = df_clean['price_lakhs'] / df_clean['area']
print("Price per sqft feature created!")
print("\n" + "="*40)

# Check for outliers in numerical columns
numerical_cols = ['bhk', 'area', 'price_lakhs', 'price_per_sqft']
print("Outlier Analysis (using IQR method):")

for col in numerical_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df_clean[(df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers ({len(outliers)/len(df_clean)*100:.2f}%)")
print("\n" + "="*40)
````

````python
print("📈 EXPLORATORY DATA ANALYSIS (EDA)")
print("=" * 40)

# Distribution of numerical variables
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Distribution of Numerical Variables', fontsize=16)

# BHK distribution
sns.countplot(data=df_clean, x='bhk', ax=axes[0,0])
axes[0,0].set_title('Distribution of BHK')

# Area distribution
sns.histplot(data=df_clean, x='area', bins=50, ax=axes[0,1])
axes[0,1].set_title('Distribution of Area')

# Price distribution
sns.histplot(data=df_clean, x='price_lakhs', bins=50, ax=axes[1,0])
axes[1,0].set_title('Distribution of Price (Lakhs)')

# Price per sqft distribution
sns.histplot(data=df_clean, x='price_per_sqft', bins=50, ax=axes[1,1])
axes[1,1].set_title('Distribution of Price per Sqft')

plt.tight_layout()
plt.show()

# Categorical variables analysis
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Categorical Variables Analysis', fontsize=16)

# Property type distribution
type_counts = df_clean['type'].value_counts()
axes[0,0].pie(type_counts.values, labels=type_counts.index, autopct='%1.1f%%')
axes[0,0].set_title('Property Type Distribution')

# Status distribution
sns.countplot(data=df_clean, y='status', ax=axes[0,1])
axes[0,1].set_title('Construction Status')

# Age distribution
sns.countplot(data=df_clean, y='age', ax=axes[1,0])
axes[1,0].set_title('Property Age')

# Price unit distribution (original)
sns.countplot(data=df_clean, x='price_unit', ax=axes[1,1])
axes[1,1].set_title('Original Price Unit Distribution')

plt.tight_layout()
plt.show()
````

````python
print("🔧 ADVANCED PANDAS OPERATIONS")
print("=" * 40)

# GROUP BY OPERATIONS
print("GROUP BY ANALYSIS:")
print("=" * 20)

# Average price by BHK
bhk_price = df_clean.groupby('bhk').agg({
    'price_lakhs': ['mean', 'median', 'std', 'count'],
    'area': 'mean',
    'price_per_sqft': 'mean'
}).round(2)
print("Price Analysis by BHK:")
print(bhk_price)
print("\n" + "="*20)

# Average price by property type and status
type_status_price = df_clean.groupby(['type', 'status']).agg({
    'price_lakhs': ['mean', 'median', 'count'],
    'area': 'mean',
    'price_per_sqft': 'mean'
}).round(2)
print("Price Analysis by Type and Status:")
print(type_status_price)
print("\n" + "="*20)

# Top 10 localities by average price
top_localities = df_clean.groupby('locality').agg({
    'price_lakhs': 'mean',
    'area': 'mean',
    'price_per_sqft': 'mean',
    'bhk': 'count'
}).round(2).nlargest(10, 'price_lakhs')
print("Top 10 Localities by Average Price:")
print(top_localities)
print("\n" + "="*20)

# MELT FUNCTION EXAMPLE
print("MELT FUNCTION EXAMPLE:")
print("=" * 20)

# Create a subset for melting
melt_subset = df_clean[['bhk', 'area', 'price_lakhs', 'price_per_sqft']].head(10)
melted_df = pd.melt(melt_subset, id_vars=['bhk'], 
                   value_vars=['area', 'price_lakhs', 'price_per_sqft'],
                   var_name='metric', value_name='value')
print("Melted DataFrame (first 10 rows):")
print(melted_df.head(10))
print("\n" + "="*20)

# FILTERING OPERATIONS
print("FILTERING EXAMPLES:")
print("=" * 20)

# Filter properties in specific price range
mid_range_properties = df_clean[(df_clean['price_lakhs'] >= 50) & (df_clean['price_lakhs'] <= 100)]
print(f"Properties in 50-100 Lakhs range: {len(mid_range_properties)}")

# Filter ready to move properties in prime locations
prime_locations = ['Andheri West', 'Bandra West', 'Powai', 'Lower Parel']
ready_prime = df_clean[(df_clean['status'] == 'Ready to move') & 
                      (df_clean['region'].isin(prime_locations))]
print(f"Ready to move properties in prime locations: {len(ready_prime)}")

# Filter new construction apartments
new_apartments = df_clean[(df_clean['type'] == 'Apartment') & 
                         (df_clean['age'] == 'New') & 
                         (df_clean['status'] == 'Under Construction')]
print(f"New under-construction apartments: {len(new_apartments)}")
print("\n" + "="*20)
````

````python
print("⚙️ FEATURE ENGINEERING")
print("=" * 40)

# Create price categories
def price_category(price):
    if price < 50:
        return 'Budget (<50L)'
    elif price < 100:
        return 'Mid-Range (50L-1Cr)'
    elif price < 200:
        return 'Premium (1-2Cr)'
    else:
        return 'Luxury (>2Cr)'

df_clean['price_category'] = df_clean['price_lakhs'].apply(price_category)
print("Price categories created!")

# Create area categories
def area_category(area):
    if area < 500:
        return 'Compact (<500)'
    elif area < 800:
        return 'Medium (500-800)'
    elif area < 1200:
        return 'Large (800-1200)'
    else:
        return 'Very Large (>1200)'

df_clean['area_category'] = df_clean['area'].apply(area_category)
print("Area categories created!")

# Create region type based on price per sqft
region_avg_pps = df_clean.groupby('region')['price_per_sqft'].mean().sort_values(ascending=False)
high_end_regions = region_avg_pps.head(10).index
mid_end_regions = region_avg_pps.iloc[10:30].index

def region_type(region):
    if region in high_end_regions:
        return 'High-End'
    elif region in mid_end_regions:
        return 'Mid-Range'
    else:
        return 'Affordable'

df_clean['region_type'] = df_clean['region'].apply(region_type)
print("Region types created!")

# Display new features
print("\nNew Features Summary:")
print(f"Price Categories: {df_clean['price_category'].value_counts().to_dict()}")
print(f"Area Categories: {df_clean['area_category'].value_counts().to_dict()}")
print(f"Region Types: {df_clean['region_type'].value_counts().to_dict()}")
print("\n" + "="*40)
````

````python
print("📊 MOVING AVERAGES AND TREND ANALYSIS")
print("=" * 40)

# Since we don't have actual dates, we'll use area as a proxy for sorting
# Create a pseudo-time series by sorting by area and calculating moving averages
df_sorted = df_clean.sort_values('area').reset_index(drop=True)

# Calculate moving averages
window_sizes = [50, 100, 200]
for window in window_sizes:
    df_sorted[f'price_ma_{window}'] = df_sorted['price_lakhs'].rolling(window=window).mean()
    df_sorted[f'pps_ma_{window}'] = df_sorted['price_per_sqft'].rolling(window=window).mean()

# Plot moving averages
fig, axes = plt.subplots(2, 1, figsize=(15, 10))
fig.suptitle('Moving Average Analysis (Sorted by Area)', fontsize=16)

# Price moving averages
axes[0].plot(df_sorted.index, df_sorted['price_lakhs'], alpha=0.3, label='Actual Price')
for window in window_sizes:
    axes[0].plot(df_sorted.index, df_sorted[f'price_ma_{window}'], 
                label=f'{window}-point MA', linewidth=2)
axes[0].set_ylabel('Price (Lakhs)')
axes[0].set_title('Price Moving Averages')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Price per sqft moving averages
axes[1].plot(df_sorted.index, df_sorted['price_per_sqft'], alpha=0.3, label='Actual PPS')
for window in window_sizes:
    axes[1].plot(df_sorted.index, df_sorted[f'pps_ma_{window}'], 
                label=f'{window}-point MA', linewidth=2)
axes[1].set_ylabel('Price per Sqft (Lakhs)')
axes[1].set_xlabel('Index (Sorted by Area)')
axes[1].set_title('Price per Sqft Moving Averages')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Moving averages calculated and plotted!")
print("\n" + "="*40)
````

````python
print("📈 TIME SERIES FORECASTING WITH ARIMA/SARIMA")
print("=" * 50)

# For time series forecasting, we need to create a time-based index
# Since we don't have actual dates, we'll use area as a proxy and create synthetic time series

from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import warnings
warnings.filterwarnings('ignore')

# Create a better synthetic time series with more data points
print("Creating Enhanced Synthetic Time Series...")

# Method 1: Use region-based aggregation for more data points
region_time_series = df_clean.groupby('region')['price_lakhs'].agg(['mean', 'count']).reset_index()
region_time_series = region_time_series[region_time_series['count'] >= 5]  # Filter regions with sufficient data
region_time_series = region_time_series.sort_values('mean').reset_index(drop=True)

# Method 2: Use BHK-based aggregation
bhk_time_series = df_clean.groupby('bhk')['price_lakhs'].mean().reset_index()
bhk_time_series = bhk_time_series.sort_values('bhk').reset_index(drop=True)

print(f"Region-based time series length: {len(region_time_series)}")
print(f"BHK-based time series length: {len(bhk_time_series)}")

# Let's use the region-based time series as it has more data points
time_series_data = region_time_series[['mean']].rename(columns={'mean': 'avg_price'})

print("\nSynthetic Time Series Data Created")
print(f"Time series length: {len(time_series_data)}")
print("\nFirst 10 periods:")
print(time_series_data.head(10))
print("\n" + "="*40)

# Stationarity test
def check_stationarity(timeseries):
    """Check stationarity using Augmented Dickey-Fuller test"""
    result = adfuller(timeseries.dropna())
    print('ADF Statistic:', result[0])
    print('p-value:', result[1])
    print('Critical Values:')
    for key, value in result[4].items():
        print(f'   {key}: {value}')
    
    if result[1] <= 0.05:
        print("✅ Series is stationary")
        return True
    else:
        print("❌ Series is not stationary")
        return False

print("Stationarity Check for Average Price:")
is_stationary = check_stationarity(time_series_data['avg_price'])
print("\n" + "="*40)

# ACF and PACF plots with adjusted lags
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# Calculate maximum allowed lags (50% of sample size)
max_lags = min(10, len(time_series_data) // 2 - 1)  # Use smaller of 10 or half sample size

plot_acf(time_series_data['avg_price'].dropna(), ax=axes[0], lags=max_lags)
axes[0].set_title(f'Autocorrelation Function (ACF) - {max_lags} lags')
plot_pacf(time_series_data['avg_price'].dropna(), ax=axes[1], lags=max_lags)
axes[1].set_title(f'Partial Autocorrelation Function (PACF) - {max_lags} lags')
plt.tight_layout()
plt.show()

# ARIMA Model
print("ARIMA MODEL FITTING")
print("=" * 20)

try:
    # Split data into train and test
    train_size = int(len(time_series_data) * 0.7)  # 70% train, 30% test
    train, test = time_series_data['avg_price'].iloc[:train_size], time_series_data['avg_price'].iloc[train_size:]
    
    print(f"Training data size: {len(train)}")
    print(f"Test data size: {len(test)}")
    
    # Fit ARIMA model with different parameters based on ACF/PACF
    # Based on ACF/PACF patterns, we'll try different orders
    arima_orders = [(1, 0, 0), (0, 1, 0), (1, 1, 0)]
    best_arima = None
    best_arima_mse = float('inf')
    
    for order in arima_orders:
        try:
            arima_model = ARIMA(train, order=order)
            arima_fit = arima_model.fit()
            
            # Forecast
            forecast_steps = len(test)
            arima_forecast = arima_fit.forecast(steps=forecast_steps)
            
            # Calculate forecast error
            mse = np.mean((test.values - arima_forecast.values) ** 2)
            
            if mse < best_arima_mse:
                best_arima_mse = mse
                best_arima = (order, arima_fit, arima_forecast)
                
            print(f"ARIMA{order} - MSE: {mse:.2f}")
            
        except Exception as e:
            print(f"ARIMA{order} failed: {e}")
    
    if best_arima:
        best_order, best_fit, best_forecast = best_arima
        
        print(f"\nBest ARIMA Model: ARIMA{best_order}")
        print("ARIMA Model Summary:")
        print(best_fit.summary())
        
        # Plot results
        plt.figure(figsize=(12, 6))
        plt.plot(range(len(train)), train.values, label='Training Data', marker='o')
        plt.plot(range(len(train), len(train) + len(test)), test.values, 
                label='Actual Test Data', color='green', marker='s')
        plt.plot(range(len(train), len(train) + len(test)), best_forecast, 
                label='ARIMA Forecast', color='red', marker='^')
        plt.title(f'Best ARIMA{best_order} Model: Price Forecasting')
        plt.xlabel('Time Period')
        plt.ylabel('Average Price (Lakhs)')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()
        
        print(f"Best ARIMA Model Mean Squared Error: {best_arima_mse:.2f}")
        
        # Calculate additional metrics
        mae = np.mean(np.abs(test.values - best_forecast.values))
        rmse = np.sqrt(best_arima_mse)
        print(f"Mean Absolute Error (MAE): {mae:.2f}")
        print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
        
    else:
        print("No ARIMA model could be successfully fitted")
    
except Exception as e:
    print(f"ARIMA modeling failed: {e}")

print("\n" + "="*40)

# SARIMA Model (Seasonal ARIMA)
print("SARIMA MODEL FITTING")
print("=" * 20)

try:
    # For SARIMA, we need to determine seasonal period
    # Since we don't have actual seasonal data, we'll assume a small seasonal period
    seasonal_period = min(4, len(train) // 3)  # Conservative seasonal period
    
    print(f"Using seasonal period: {seasonal_period}")
    
    # Try different SARIMA configurations
    sarima_orders = [
        (1, 1, 1, seasonal_period),
        (1, 0, 0, seasonal_period),
        (0, 1, 1, seasonal_period)
    ]
    
    best_sarima = None
    best_sarima_mse = float('inf')
    
    for order in sarima_orders:
        try:
            p, d, q, s = order
            sarima_model = SARIMAX(train, order=(p, d, q), seasonal_order=(p, d, q, s))
            sarima_fit = sarima_model.fit(disp=False)
            
            # Forecast
            sarima_forecast = sarima_fit.forecast(steps=len(test))
            
            # Calculate forecast error
            mse = np.mean((test.values - sarima_forecast.values) ** 2)
            
            if mse < best_sarima_mse:
                best_sarima_mse = mse
                best_sarima = (order, sarima_fit, sarima_forecast)
                
            print(f"SARIMA{order} - MSE: {mse:.2f}")
            
        except Exception as e:
            print(f"SARIMA{order} failed: {e}")
    
    if best_sarima:
        best_order, best_fit, best_forecast = best_sarima
        
        print(f"\nBest SARIMA Model: SARIMA{best_order}")
        print("SARIMA Model Summary:")
        print(best_fit.summary())
        
        # Plot results
        plt.figure(figsize=(12, 6))
        plt.plot(range(len(train)), train.values, label='Training Data', marker='o')
        plt.plot(range(len(train), len(train) + len(test)), test.values, 
                label='Actual Test Data', color='green', marker='s')
        plt.plot(range(len(train), len(train) + len(test)), best_forecast, 
                label='SARIMA Forecast', color='red', marker='^')
        plt.title(f'Best SARIMA{best_order} Model: Price Forecasting')
        plt.xlabel('Time Period')
        plt.ylabel('Average Price (Lakhs)')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()
        
        print(f"Best SARIMA Model Mean Squared Error: {best_sarima_mse:.2f}")
        
        # Calculate additional metrics
        mae = np.mean(np.abs(test.values - best_forecast.values))
        rmse = np.sqrt(best_sarima_mse)
        print(f"Mean Absolute Error (MAE): {mae:.2f}")
        print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
        
        # Compare ARIMA vs SARIMA
        if best_arima:
            print(f"\n📊 MODEL COMPARISON:")
            print(f"ARIMA{best_order} MSE: {best_arima_mse:.2f}")
            print(f"SARIMA{best_order} MSE: {best_sarima_mse:.2f}")
            
            if best_arima_mse < best_sarima_mse:
                print("✅ ARIMA performs better for this data")
            else:
                print("✅ SARIMA performs better for this data")
        
    else:
        print("No SARIMA model could be successfully fitted")
    
except Exception as e:
    print(f"SARIMA modeling failed: {e}")

print("\n" + "="*40)

# Alternative Forecasting Approach: Simple Models
print("🔮 ALTERNATIVE FORECASTING METHODS")
print("=" * 30)

# Simple moving average forecast
def moving_average_forecast(data, window):
    """Simple moving average forecast"""
    return data.rolling(window=window).mean().iloc[-1]

# Simple exponential smoothing
def exponential_smoothing_forecast(data, alpha=0.3):
    """Simple exponential smoothing forecast"""
    forecast = [data.iloc[0]]
    for i in range(1, len(data)):
        forecast.append(alpha * data.iloc[i] + (1 - alpha) * forecast[i-1])
    return forecast[-1]

# Try simple methods
try:
    # Last value (naive forecast)
    naive_forecast = train.iloc[-1]
    naive_mse = np.mean((test.values - naive_forecast) ** 2)
    
    # Moving average (window=3)
    ma_forecast = moving_average_forecast(train, window=3)
    ma_mse = np.mean((test.values - ma_forecast) ** 2)
    
    # Exponential smoothing
    es_forecast = exponential_smoothing_forecast(train, alpha=0.3)
    es_mse = np.mean((test.values - es_forecast) ** 2)
    
    print("Simple Forecasting Methods Comparison:")
    print(f"Naive Forecast (last value): MSE = {naive_mse:.2f}")
    print(f"Moving Average (window=3): MSE = {ma_mse:.2f}")
    print(f"Exponential Smoothing (alpha=0.3): MSE = {es_mse:.2f}")
    
    # Plot comparison of simple methods
    plt.figure(figsize=(12, 6))
    plt.plot(range(len(train)), train.values, label='Training Data')
    plt.plot(range(len(train), len(train) + len(test)), test.values, 
            label='Actual Test Data', color='green', linewidth=2)
    
    # Add forecast lines
    plt.axhline(y=naive_forecast, color='red', linestyle='--', 
               label=f'Naive Forecast (MSE: {naive_mse:.2f})')
    plt.axhline(y=ma_forecast, color='blue', linestyle='--', 
               label=f'MA Forecast (MSE: {ma_mse:.2f})')
    plt.axhline(y=es_forecast, color='orange', linestyle='--', 
               label=f'ES Forecast (MSE: {es_mse:.2f})')
    
    plt.title('Simple Forecasting Methods Comparison')
    plt.xlabel('Time Period')
    plt.ylabel('Average Price (Lakhs)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
    
except Exception as e:
    print(f"Simple forecasting methods failed: {e}")

print("\n" + "="*50)
print("✅ TIME SERIES ANALYSIS COMPLETED!")
print("=" * 50)
````
